**Question 1: Debugging (Compound Logic)**

The following script is designed to flag an expense report for manual audit if it meets *either* of these conditions:
1. The expense is over $10,000.

2. The expense is over $5,000 AND the category is "Miscellaneous".

**Group Discussion:** Look at the code below. Why is it flagging a $6,000 "Travel" expense for an audit?

**Fix the logical errors, syntax errors, and add comments explaining your fixes.**

```python
# Convert the input from text to a number so it can be compared correctly.
expense_amount = float(input("Enter expense amount: "))
category = input("Enter expense category: ").strip()

# Use == for comparison and parentheses to make the compound logic clear.
if expense_amount > 10000 or (
    expense_amount > 5000 and category.lower() == "miscellaneous"
):
    print("ALERT: Flagged for manual audit!")
else:
    print("Auto-approved.")
```

**Question 2: Segregation of Duties (SoD) Checker**

In an ERP system, a user should not be able to both create a vendor and pay a vendor.

Write a program that:
* Asks the user to input their Primary System Role.
* Asks the user to input their Secondary System Role.
* Checks for a toxic combination. If one role is "Vendor Management" and the other is "Accounts Payable" (in any order), print: `"CRITICAL RISK: SoD Conflict Detected."`
* Otherwise, print: `"Access combinations approved."`

*Hint: You will need to use `and` / `or` operators to account for the order in which the user types the roles.*

```python
primary_role = input("Enter your Primary System Role: ").strip().lower()
secondary_role = input("Enter your Secondary System Role: ").strip().lower()

conflict = (
    primary_role == "vendor management"
    and secondary_role == "accounts payable"
) or (
    primary_role == "accounts payable"
    and secondary_role == "vendor management"
)

if conflict:
    print("CRITICAL RISK: SoD Conflict Detected.")
else:
    print("Access combinations approved.")
```

**Question 3: Nested Logic for Asset Depreciation**

Your firm has a new tax rule for software and hardware assets.
* If the asset is "Hardware":
  * and it costs 5,000 or more, it gets a $1000 deprecation bonus.
  
  * And it costs less than 5,000, it gets a $250 depreciation bonus.

* If the asset is "Software":
  * It always gets a flat $500 depreciation bonus regardless of cost.
* For any other asset type, print "Invalid asset class."

Write a program that takes the **asset type** and **cost** as inputs, uses nested `if`/`elif` statements to determine the bonus, and prints the result.

```python
asset_type = input("Enter the asset type: ").strip().lower()
cost = float(input("Enter the asset cost: "))

if asset_type == "hardware":
    if cost >= 5000:
        bonus = 1000
    else:
        bonus = 250
    print(f"Depreciation bonus: ${bonus:.2f}")

elif asset_type == "software":
    bonus = 500
    print(f"Depreciation bonus: ${bonus:.2f}")

else:
    print("Invalid asset class.")
```

**Question 4: The Payroll Audit Challenge (Group Challenge)**

Your group needs to write a script to calculate weekly payroll for hourly data entry clerks, factoring in overtime and a standard tax withholding.

**Requirements:**
* Ask for `hours_worked` and `hourly_rate`.
* Regular pay applies up to 40 hours.
* Overtime pay (1.5x the hourly rate) applies to any hours worked *over* 40.
* After calculating the Gross Pay (Regular + Overtime), deduct a flat 20% for taxes.
* Print a clean "Pay Stub" showing: Gross Pay, Taxes Withheld, and Net Pay (formatted to two decimal places).

```python
# Ask for inputs
hours_worked = float(input("Enter hours worked: "))
hourly_rate = float(input("Enter hourly rate: $"))

# Calculate Gross Pay
if hours_worked <= 40:
    gross_pay = hours_worked * hourly_rate
else:
    regular_pay = 40 * hourly_rate
    overtime_hours = hours_worked - 40
    overtime_pay = overtime_hours * hourly_rate * 1.5
    gross_pay = regular_pay + overtime_pay

# Calculate Taxes and Net Pay
taxes_withheld = gross_pay * 0.20
net_pay = gross_pay - taxes_withheld

# Print the Pay Stub
print("\n--- Pay Stub ---")
print(f"Gross Pay:       ${gross_pay:.2f}")
print(f"Taxes Withheld:  ${taxes_withheld:.2f}")
print(f"Net Pay:         ${net_pay:.2f}")
```
